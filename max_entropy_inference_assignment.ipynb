# **Statistical Mechanics of the Quantum Oscillator**

1. The Partition FunctionThe canonical partition function $Z$ is defined as the sum over all possible microstates of the Boltzmann factor:$$Z = \sum_{n=0}^{\infty} e^{-\beta E_n} = \sum_{n=0}^{\infty} e^{-\beta \hbar \omega \left(n + \frac{1}{2}\right)}$$We can factor out the constant ground-state energy term ($n=0$):$$Z = e^{-\frac{1}{2}\beta \hbar \omega} \sum_{n=0}^{\infty} \left(e^{-\beta \hbar \omega}\right)^n$$The summation is a geometric series of the form $\sum_{n=0}^{\infty} x^n$ where $x = e^{-\beta \hbar \omega}$. Since $\beta > 0$ and $\hbar \omega > 0$, the absolute value $|x| < 1$, ensuring the series cleanly converges to $\frac{1}{1-x}$. Substituting $x$ back into the formula yields:$$Z = \frac{e^{-\frac{1}{2}\beta \hbar \omega}}{1 - e^{-\beta \hbar \omega}}$$

2. Thermodynamic PotentialsHelmholtz Free Energy ($F$)Using the fundamental relation $F = -k_B T \ln Z = -\frac{1}{\beta} \ln Z$:$$F = -\frac{1}{\beta} \ln \left( \frac{e^{-\frac{1}{2}\beta \hbar \omega}}{1 - e^{-\beta \hbar \omega}} \right) = -\frac{1}{\beta} \left[ -\frac{1}{2}\beta \hbar \omega - \ln\left(1 - e^{-\beta \hbar \omega}\right) \right]$$$$F = \frac{1}{2}\hbar \omega + k_B T \ln\left(1 - e^{-\beta \hbar \omega}\right)$$Internal Energy ($U$)The internal energy can be derived using the relation $U = -\frac{\partial \ln Z}{\partial \beta}$:$$\ln Z = -\frac{1}{2}\beta \hbar \omega - \ln\left(1 - e^{-\beta \hbar \omega}\right)$$$$U = -\frac{\partial}{\partial \beta} \left[ -\frac{1}{2}\beta \hbar \omega - \ln\left(1 - e^{-\beta \hbar \omega}\right) \right] = \frac{1}{2}\hbar \omega + \frac{1}{1 - e^{-\beta \hbar \omega}} \cdot \left(-e^{-\beta \hbar \omega}\right) \cdot (-\hbar \omega)$$$$U = \frac{1}{2}\hbar \omega + \frac{\hbar \omega e^{-\beta \hbar \omega}}{1 - e^{-\beta \hbar \omega}} = \frac{1}{2}\hbar \omega + \frac{\hbar \omega}{e^{\beta \hbar \omega} - 1}$$

3. Entropy ($S$)Using the thermodynamic identity $S = -\frac{\partial F}{\partial T}$:$$S = -\frac{\partial}{\partial T} \left[ \frac{1}{2}\hbar \omega + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) \right]$$$$S = -k_B \ln\left(1 - e^{-\beta \hbar \omega}\right) - k_B T \cdot \frac{1}{1 - e^{-\beta \hbar \omega}} \cdot \left(-e^{-\beta \hbar \omega}\right) \cdot \left(\frac{\hbar \omega}{k_B T^2}\right)$$$$S = -k_B \ln\left(1 - e^{-\beta \hbar \omega}\right) + \frac{\hbar \omega}{T} \frac{1}{e^{\beta \hbar \omega} - 1}$$Verification of Third Law ($\lim_{T \to 0} S = 0$)As $T \to 0$, $\beta = \frac{1}{k_B T} \to \infty$.The exponential term $e^{-\beta \hbar \omega} \to 0$, so $\ln(1 - e^{-\beta \hbar \omega}) \to \ln(1) = 0$.The second term becomes dominated by the denominator: $\lim_{\beta \to \infty} \frac{\beta k_B \hbar \omega}{e^{\beta \hbar \omega} - 1} = 0$.Thus, $\lim_{T \to 0} S = 0$, satisfying the Nernst heat theorem (Third Law of Thermodynamics).

4. Heat Capacity and Hyperbolic FormThe constant-volume heat capacity is given by $C_V = \frac{\partial U}{\partial T} = \frac{\partial U}{\partial \beta} \frac{\partial \beta}{\partial T} = -k_B \beta^2 \frac{\partial U}{\partial \beta}$:$$\frac{\partial U}{\partial \beta} = \frac{\partial}{\partial \beta} \left( \frac{1}{2}\hbar \omega + \frac{\hbar \omega}{e^{\beta \hbar \omega} - 1} \right) = -\frac{(\hbar \omega)^2 e^{\beta \hbar \omega}}{(e^{\beta \hbar \omega} - 1)^2}$$$$C_V = -k_B \beta^2 \left( -\frac{(\hbar \omega)^2 e^{\beta \hbar \omega}}{(e^{\beta \hbar \omega} - 1)^2} \right) = k_B (\beta \hbar \omega)^2 \frac{e^{\beta \hbar \omega}}{(e^{\beta \hbar \omega} - 1)^2}$$To rewrite this in hyperbolic form, multiply the numerator and denominator by $e^{-\beta \hbar \omega}$:$$C_V = k_B (\beta \hbar \omega)^2 \frac{1}{(e^{\frac{\beta \hbar \omega}{2}} - e^{-\frac{\beta \hbar \omega}{2}})^2}$$Using the identity $\sinh(x) = \frac{e^x - e^{-x}}{2} \implies e^x - e^{-x} = 2\sinh(x)$, let $x = \frac{\beta \hbar \omega}{2}$:$$C_V = k_B (\beta \hbar \omega)^2 \frac{1}{\left(2 \sinh\left(\frac{\beta \hbar \omega}{2}\right)\right)^2} = k_B \left(\frac{\beta \hbar \omega}{2}\right)^2 \frac{1}{\sinh^2\left(\frac{\beta \hbar \omega}{2}\right)}$$$$C_V = k_B \left(\frac{\beta \hbar \omega}{2}\right)^2 \text{csch}^2\left(\frac{\beta \hbar \omega}{2}\right)$$

5. Asymptotic Behavior (High-Temperature Limit)In the high-temperature limit where $T \to \infty$, we have $\beta \to 0$, which means $x = \frac{\beta \hbar \omega}{2} \ll 1$.Using the small-argument approximation for hyperbolic sine, $\sinh(x) \approx x$:$$C_V \approx k_B x^2 \left(\frac{1}{x}\right)^2 = k_B$$Conclusion: Yes, this perfectly matches the classical equipartition theorem prediction. A classical 1D harmonic oscillator has two quadratic degrees of freedom (kinetic and potential energy), each contributing $\frac{1}{2}k_B$ to the heat capacity, totaling $C_V = k_B$.

# **The Maximum Entropy Distribution under Mean and Variance Constraints**

To find the probability density function $f(x)$ that maximizes the continuous differential entropy $H(f) = -\int_{-\infty}^{\infty} f(x) \ln f(x) \, dx$, subject to normalization, mean, and variance constraints, we construct the functional using Lagrange multipliers $\lambda_0, \lambda_1, \lambda_2$:$$\mathcal{L}(f, \lambda_0, \lambda_1, \lambda_2) = -\int_{-\infty}^{\infty} f(x) \ln f(x) \, dx - \lambda_0 \left( \int_{-\infty}^{\infty} f(x) \, dx - 1 \right) - \lambda_1 \left( \int_{-\infty}^{\infty} x f(x) \, dx - \mu \right) - \lambda_2 \left( \int_{-\infty}^{\infty} (x-\mu)^2 f(x) \, dx - \sigma^2 \right)$$Taking the functional derivative with respect to $f(x)$ and setting it to zero:$$\frac{\delta \mathcal{L}}{\delta f(x)} = - \ln f(x) - 1 - \lambda_0 - \lambda_1 x - \lambda_2 (x - \mu)^2 = 0$$$$\ln f(x) = -(1 + \lambda_0) - \lambda_1 x - \lambda_2 (x - \mu)^2 \implies f(x) = C e^{-\lambda_1 x - \lambda_2 (x-\mu)^2}$$By shifting the coordinate system around the center of the constraints, the linear term $\lambda_1 x$ must vanish ($\lambda_1 = 0$) to keep the distribution symmetrically centered precisely at the given mean $\mu$. Thus:$$f(x) = C e^{-\lambda_2 (x-\mu)^2}$$We evaluate the constants using our normalization and variance properties:Normalization: $\int_{-\infty}^{\infty} C e^{-\lambda_2 (x-\mu)^2} \, dx = C \sqrt{\frac{\pi}{\lambda_2}} = 1 \implies C = \sqrt{\frac{\lambda_2}{\pi}}$Variance: $\int_{-\infty}^{\infty} (x-\mu)^2 \left(\sqrt{\frac{\lambda_2}{\pi}} e^{-\lambda_2 (x-\mu)^2}\right) \, dx = \frac{1}{2\lambda_2} = \sigma^2 \implies \lambda_2 = \frac{1}{2\sigma^2}$Substituting $\lambda_2$ back into the expression for $C$ yields $C = \frac{1}{\sqrt{2\pi\sigma^2}}$. Reassembling the equation proves that the MaxEnt distribution is the Gaussian distribution:$$f(x) = \frac{1}{\sqrt{2\pi\sigma^2}} e^{-\frac{(x-\mu)^2}{2\sigma^2}}$$

# **Multiclass Logistic Regression: Automatic Classification of Emails**

To model the routing probabilities using maximum entropy framework principles (equivalent to multiclass logistic regression / Softmax), the conditional probability of routing an email to destination $y \in \{Y_1, Y_2\}$ given an observed binary feature vector $\mathbf{x} = [x_1, x_2, x_3]^T$ is formulated as:$$P(y | \mathbf{x}) = \frac{e^{\mathbf{w}_y \cdot \mathbf{x} + b_y}}{\sum_{j \in \{Y_1, Y_2\}} e^{\mathbf{w}_j \cdot \mathbf{x} + b_j}}$$Because the problem specifies a binary choice ($y \in \{Y_1, Y_2\}$ where $Y_1 = \text{Finance}$, $Y_2 = \text{Technical}$), we can fix $Y_2$ as the base reference class ($\mathbf{w}_{Y_2} = \mathbf{0}, b_{Y_2} = 0$). This reduces the system to a single parameter vector $\mathbf{w} = [w_1, w_2, w_3]^T$ and bias $b$ for the Finance category:$$P(Y_1 | \mathbf{x}) = \frac{e^{\mathbf{w} \cdot \mathbf{x} + b}}{1 + e^{\mathbf{w} \cdot \mathbf{x} + b}} = \sigma(\mathbf{w} \cdot \mathbf{x} + b), \quad P(Y_2 | \mathbf{x}) = \frac{1}{1 + e^{\mathbf{w} \cdot \mathbf{x} + b}}$$Let's evaluate the operational feature states corresponding to each scenario:Scenario A (Pure Financial): $\mathbf{x}_A = [1, 0, 0]^T \implies P(Y_1 | \mathbf{x}_A) = \sigma(w_1 + b)$Scenario B (Pure Access): $\mathbf{x}_B = [0, 1, 0]^T \implies P(Y_1 | \mathbf{x}_B) = \sigma(w_2 + b)$Scenario C (Pure Technical): $\mathbf{x}_C = [0, 0, 1]^T \implies P(Y_1 | \mathbf{x}_C) = \sigma(w_3 + b)$Scenario D (Access Technical): $\mathbf{x}_D = [0, 1, 1]^T \implies P(Y_1 | \mathbf{x}_D) = \sigma(w_2 + w_3 + b)$Scenario E (Billing Access): $\mathbf{x}_E = [1, 1, 0]^T \implies P(Y_1 | \mathbf{x}_E) = \sigma(w_1 + w_2 + b)$

**Solving for Parameters Using Constraints**

Step 1: Resolve Constraint 1Constraint 1 states:

$E[x_1 \cdot \mathbb{I}(y = Y_2)] = 0.01$.Expanding this joint expectation using law of total probability across the scenarios containing $x_1=1$ (Scenarios A and E):$$P(\text{Scenario A}) \cdot 1 \cdot P(Y_2 | \mathbf{x}_A) + P(\text{Scenario E}) \cdot 1 \cdot P(Y_2 | \mathbf{x}_E) = 0.01$$$$0.30 \cdot [1 - \sigma(w_1 + b)] + 0.10 \cdot [1 - \sigma(w_1 + w_2 + b)] = 0.01$$

Step 2: Resolve Constraint 2Constraint 2 states: $E[x_2 \cdot \mathbb{I}(y = Y_1)] = 0.20$.Expanding across scenarios containing $x_2=1$ (Scenarios B, D, and E):$$P(\text{Scenario B}) \cdot \sigma(w_2 + b) + P(\text{Scenario D}) \cdot \sigma(w_2 + w_3 + b) + P(\text{Scenario E}) \cdot \sigma(w_1 + w_2 + b) = 0.20$$$$0.20 \cdot \sigma(w_2 + b) + 0.15 \cdot \sigma(w_2 + w_3 + b) + 0.10 \cdot \sigma(w_1 + w_2 + b) = 0.20$$Step 3: Resolve Constraint 3Constraint 3 states: $E[x_3 \cdot \mathbb{I}(y = Y_2)] = 0.36$.Expanding across scenarios containing $x_3=1$ (Scenarios C and D):$$P(\text{Scenario C}) \cdot [1 - \sigma(w_3 + b)] + P(\text{Scenario D}) \cdot [1 - \sigma(w_2 + w_3 + b)] = 0.36$$$$0.25 \cdot [1 - \sigma(w_3 + b)] + 0.15 \cdot [1 - \sigma(w_2 + w_3 + b)] = 0.36$$Notice from the total probabilities of Scenarios C and D that $0.25 + 0.15 = 0.40$. Rewriting the equation:$$0.40 - 0.25\sigma(w_3 + b) - 0.15\sigma(w_2 + w_3 + b) = 0.36 \implies 0.25\sigma(w_3 + b) + 0.15\sigma(w_2 + w_3 + b) = 0.04$$Since the maximum outputs must respect these boundary expressions, numerical optimization algorithms (such as Iterative Scaling or Newton-Raphson) are universally used to extract exact weights. However, we can approximate the underlying log-odds directly from these explicit conditional ratios:From Step 3, if $\sigma(w_3+b)$ and $\sigma(w_2+w_3+b)$ are roughly close, their average conditional probability is $\frac{0.04}{0.40} = 0.10$. This implies a baseline routing to Technical support of $90\%$ when a tech keyword is present.From Step 1, the probability of an email containing a financial feature going to Technical support is $\frac{0.01}{0.40} = 0.025$, meaning $\sigma(w_1+b) \approx 0.975$.This directly generates the log-odds mapping required for the automated classification routing.

# **Asset Pricing and Portfolio Optimization**

We seek the continuous probability density function $p(x)$ for the asset price $x \in (0, \infty)$ that maximizes the differential entropy:$$\max_{p} -\int_{0}^{\infty} p(x) \ln p(x) \, dx$$Subject to the following market constraints:Normalization: $\int_{0}^{\infty} p(x) \, dx = 1$Risk-Free Growth: $\int_{0}^{\infty} x p(x) \, dx = F_0$Market Volatility: $\int_{0}^{\infty} x^2 p(x) \, dx = V_0$Asymmetric Market Sentiment: $\int_{0}^{\infty} (\ln x) p(x) \, dx = S_0$The MaxEnt Functional FormUsing Lagrange multipliers $\lambda_0, \lambda_1, \lambda_2, \lambda_3$, the optimization objective function expands to:$$\mathcal{L} = -\int_{0}^{\infty} p(x) \ln p(x) \, dx - \lambda_0 \left(\int_{0}^{\infty} p(x)dx - 1\right) - \lambda_1 \left(\int_{0}^{\infty} x p(x)dx - F_0\right) - \lambda_2 \left(\int_{0}^{\infty} x^2 p(x)dx - V_0\right) - \lambda_3 \left(\int_{0}^{\infty} (\ln x) p(x)dx - S_0\right)$$Taking the functional derivative with respect to $p(x)$ and setting it to 0:$$\frac{\delta \mathcal{L}}{\delta p(x)} = - \ln p(x) - 1 - \lambda_0 - \lambda_1 x - \lambda_2 x^2 - \lambda_3 \ln x = 0$$Solving for $p(x)$:$$\ln p(x) = -(1 + \lambda_0) - \lambda_1 x - \lambda_2 x^2 - \lambda_3 \ln x$$$$p(x) = e^{-(1+\lambda_0)} \cdot e^{-\lambda_3 \ln x} \cdot e^{-\lambda_1 x - \lambda_2 x^2}$$$$p(x) = \frac{1}{Z} x^{-\lambda_3} e^{-\lambda_1 x - \lambda_2 x^2}$$Where $Z = e^{1+\lambda_0} = \int_{0}^{\infty} x^{-\lambda_3} e^{-\lambda_1 x - \lambda_2 x^2} \, dx$ serves as the normalizing partition function.

# **Distribution Analysis**

The resulting risk-neutral probability density function belongs to the unifying Exponential Family.Depending on the specific numerical values of the empirical market inputs ($F_0, V_0, S_0$), this structural family elegantly interpolates between classical foundational risk profiles:If $\lambda_2 = 0$, the structure simplifies exactly to a Gamma distribution, naturally capturing skewness and zero-bound price dynamics.If the constraints closely mimic symmetric variances around the forward rate with a balanced logarithmic structure, the distribution transitions toward a traditional Log-Normal framework.This yields the least-biased, most robust asset pricing distribution fully consistent with the observed multi-option market parameters.
